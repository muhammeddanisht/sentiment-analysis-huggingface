```python
from transformers import pipeline
```

```python
classifier = pipeline("sentiment-analysis")
```

```python
result = classifier("I love learning AI")
print(result)
```

```python
result = classifier("I hate Mondays")
print(result)
```

```python
result = classifier("I love to travel")
print(result)
```

```python
sentences = [
    "I love to travel",
    "I hate being stuck at home",
    "Switzerland is my dream destination",
    "I am scared of failing",
    "I will achieve financial independence"
]

for sentence in sentences:
    result = classifier(sentence)
    label = result[0]['label']
    score = round(result[0]['score'] * 100, 2)
    print(f"{sentence} → {label} ({score}%)")
```
